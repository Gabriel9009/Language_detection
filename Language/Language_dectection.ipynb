```python
from datasets import load_dataset
ds = load_dataset("papluca/language-identification")
ds['train'].to_csv("language_data.csv", index=False)
```

```python
import pandas as pd           # for working with data tables
import numpy as np            # for math operations
import matplotlib.pyplot as plt
import seaborn as sns         # for pretty charts
import re                     # for text cleaning (regular expressions)
```

```python
from sklearn.model_selection import train_test_split   # splits data into train/test
from sklearn.feature_extraction.text import TfidfVectorizer  # converts text to numbers
from sklearn.linear_model import LogisticRegression    # our main ML model
from sklearn.naive_bayes import MultinomialNB          # a second model to compare
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score
)
import warnings
warnings.filterwarnings("ignore")  # hide minor warnings so output stays clean

print("✅ All libraries imported successfully!\n")
```

```python
df_raw = pd.read_csv("language_data.csv")
```

```python
df_raw
```

```python
target_langs = ['en', 'es', 'sw', 'zh']
df = df_raw[df_raw['labels'].isin(target_langs)].reset_index(drop=True)
```

```python
df = df.groupby('labels').sample(n=2000, random_state=42).reset_index(drop=True)
```

```python
df
```

```python
def clean_text(text):
    """
    Cleans a single text string.
    - Removes extra spaces
    - Lowercases English/Spanish/Swahili text
    - Keeps Chinese characters untouched
    """
    # Remove leading/trailing whitespace
    text = text.strip()

    # Only lowercase if text doesn't contain Chinese characters
    # (Chinese doesn't have uppercase/lowercase)
    if not any('\u4e00' <= c <= '\u9fff' for c in text):
        text = text.lower()

    # Remove multiple spaces → single space
    text = re.sub(r'\s+', ' ', text)

    return text
```

```python
print("\n🧹 Cleaning text...")
df["clean_text"] = df["text"].apply(clean_text)
print("✅ Text cleaned!")
print("\n📝 Before vs After cleaning example:")
print(f"   Before: {df['text'][0]}")
print(f"   After:  {df['clean_text'][0]}")
```

```python
print("\n🔢 Converting text to numbers using TF-IDF...")

# TfidfVectorizer looks at combinations of 1-4 characters
# This captures language patterns without needing word-level tokenization
vectorizer = TfidfVectorizer(
    analyzer='char_wb',   # look at CHARACTER patterns (not just words)
    ngram_range=(1, 4),   # look at 1, 2, 3, and 4 character combinations
    max_features=10000,   # keep the top 10,000 most useful patterns
    sublinear_tf=True     # use log scaling to reduce impact of very frequent chars
)

X = vectorizer.fit_transform(df["clean_text"])  # the text features (input)
y = df["labels"]                                  # the language labels (output)

print(f"✅ Text converted to numbers!")
print(f"   Shape of feature matrix: {X.shape}")
print(f"   (That's {X.shape[0]} text samples × {X.shape[1]} character patterns)")
```

```python
print("\n✂️  Splitting data into train/test sets...")

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,       # 20% goes to testing
    random_state=42,     # fixed seed so you get the same split every time
    stratify=y           # ensure each language is equally represented in splits
)

print(f"✅ Data split!")
print(f"   Training samples: {X_train.shape[0]}")
print(f"   Testing samples:  {X_test.shape[0]}")
```

```python
print("\n🤖 Training models...")

# --- Model 1: Logistic Regression ---
print("   Training Logistic Regression...")
lr_model = LogisticRegression(
    max_iter=1000,    # how many times to try improving
    random_state=42,
    C=10              # how strongly to fit the training data
)
lr_model.fit(X_train, y_train)
print("   ✅ Logistic Regression trained!")

# --- Model 2: Naive Bayes ---
print("   Training Naive Bayes...")
nb_model = MultinomialNB(alpha=0.1)  # alpha is a smoothing factor
nb_model.fit(X_train, y_train)
print("   ✅ Naive Bayes trained!")

```

```python
print("\n" + "="*60)
print("📊 MODEL EVALUATION RESULTS")
print("="*60)

# --- Evaluate Logistic Regression ---
lr_preds = lr_model.predict(X_test)
lr_accuracy = accuracy_score(y_test, lr_preds)

print(f"\n🔵 LOGISTIC REGRESSION")
print(f"   Overall Accuracy: {lr_accuracy:.4f} ({lr_accuracy*100:.2f}%)")
print(f"\n   Detailed Report:")
print(classification_report(y_test, lr_preds,
                             target_names=["English(en)", "Mandarin(zh)", "Spanish(es)", "Swahili(sw)"],
                             labels=["en", "zh", "es", "sw"]))

# --- Evaluate Naive Bayes ---
nb_preds = nb_model.predict(X_test)
nb_accuracy = accuracy_score(y_test, nb_preds)

print(f"\n🟠 NAIVE BAYES")
print(f"   Overall Accuracy: {nb_accuracy:.4f} ({nb_accuracy*100:.2f}%)")
print(f"\n   Detailed Report:")
print(classification_report(y_test, nb_preds,
                             target_names=["English(en)", "Mandarin(zh)", "Spanish(es)", "Swahili(sw)"],
                             labels=["en", "zh", "es", "sw"]))
```

```python
# Pick the best model
best_model = lr_model if lr_accuracy >= nb_accuracy else nb_model
best_name = "Logistic Regression" if lr_accuracy >= nb_accuracy else "Naive Bayes"
print(f"\n🏆 Best model: {best_name} with {max(lr_accuracy, nb_accuracy)*100:.2f}% accuracy")
```

```python

print("🎯 TRY THE MODEL ON NEW TEXT")
language_names = {
    "en": "🇬🇧 English",
    "es": "🇪🇸 Spanish",
    "sw": "🇰🇪 Swahili",
    "zh": "🇨🇳 Mandarin"
}

def predict_language(text, model=best_model):
    """
    Takes a text string and returns the predicted language.
    Shows confidence % for each possible language.
    """
    cleaned = clean_text(text)                          # Step 1: clean it
    vectorized = vectorizer.transform([cleaned])         # Step 2: convert to numbers
    prediction = model.predict(vectorized)[0]            # Step 3: predict
    probabilities = model.predict_proba(vectorized)[0]  # Step 4: get confidence

    print(f"\n  📝 Input text : \"{text}\"")
    print(f"  🔍 Prediction : {language_names[prediction]}")
    print(f"  📊 Confidence breakdown:")
    for lang_code, prob in zip(model.classes_, probabilities):
        bar = "█" * int(prob * 30)
        print(f"      {language_names[lang_code]:15s} {bar:<30} {prob*100:5.1f}%")

    return prediction

# Test on some new sentences
test_sentences = [
    "Hello, how are you doing today?",
    "¿Cómo estás hoy? Espero que muy bien.",
    "Habari yako? Niko sawa sana, asante.",
    "你好，今天天气怎么样？",
    "The artificial intelligence revolution is here.",
    "La inteligencia artificial está cambiando el mundo.",
]

print("\nTesting on new sentences the model has never seen before:\n")
for sentence in test_sentences:
    predict_language(sentence)

print("\n\n" + "="*60)
print("✅ PROJECT COMPLETE!")
print("="*60)
```
